Cleaning STARTS HERE

```python
# Original data ko safe rakhne ke liye copy
clear_df = df.copy()
print("original rows:", df.shape[0])
print("cleandataset rows:", df.shape[1])
```

```python
# Unnecessary column remove karna
# Remove the extra empty column only if it exists
clear_df = clear_df.drop(columns=[col for col in ["Unnamed: 22"] if col in clear_df.columns])

print("Columns after removing unnecessary column:", clear_df.shape[1])

```

```python
# Date ko proper date format mein convert karna
clear_df["Date"] = pd.to_datetime(
    clear_df["Date"],
    format="%m-%d-%y",
    errors="coerce"
)

print(clear_df["Date"].dtype)
```

```python
print("Rows:", clear_df.shape[0])
print("Columns:", clear_df.shape[1])

print("\nMissing Values:")
print(clear_df.isnull().sum()[clear_df.isnull().sum() > 0])

print("\nDuplicate Rows:", clear_df.duplicated().sum())
```

```python
print(clear_df["Date"].dtype)
```

```python
# Missing values ka percentage
missing = clear_df.isnull().sum()
missing_percent = (missing / len(clear_df)) * 100

missing_report = pd.DataFrame({
    "Missing Count": missing,
    "Missing %": missing_percent.round(2)
})

missing_report = missing_report[missing_report["Missing Count"] > 0]
missing_report.sort_values("Missing %", ascending=False)
```

```python
# Text columns se extra spaces remove karna

text_columns = clear_df.select_dtypes(include=["object"]).columns

for col in text_columns:
    clear_df[col] = clear_df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Text columns cleaned successfully!")
```

```python
# Amount ko numeric format mein convert karna

clear_df["Amount"] = pd.to_numeric(
    clear_df["Amount"],
    errors="coerce"
)

print("Amount datatype:", clear_df["Amount"].dtype)
```

```python
# Missing currency ko INR se fill karna

clear_df["currency"] = clear_df["currency"].fillna("INR")

print("Currency cleaning completed!")
```

```python
# Missing location values ko Unknown se fill karna

for col in ["ship-city", "ship-state", "ship-country"]:
    clear_df[col] = clear_df[col].fillna("Unknown")

print("Location cleaning completed!")
```

```python
# Missing courier status

clear_df["Courier Status"] = clear_df["Courier Status"].fillna("Unknown")

print("Courier Status cleaning completed!")
```

```python
# Missing promotion IDs ko identify karna

clear_df["promotion-ids"] = clear_df["promotion-ids"].fillna("No Promotion")

print("Promotion IDs cleaned successfully!")
```

```python
# Missing Fulfilled By values ko handle karna

clear_df["fulfilled-by"] = clear_df["fulfilled-by"].fillna("Unknown")

print("Fulfilled By cleaned successfully!")
```

```python
#cleaning verify karo
print("Rows:", clear_df.shape[0])
print("Columns:", clear_df.shape[1])

print("\nRemaining Missing Values:")
print(clear_df.isnull().sum()[clear_df.isnull().sum() > 0])

print("\nDuplicate Rows:", clear_df.duplicated().sum())
```

```python
clear_df.to_csv(
    "../data/cleaned/Amazon_Sale_Report_Cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
